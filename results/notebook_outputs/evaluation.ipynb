```python
import os
import json
import asyncio
from openai import AsyncOpenAI
from tqdm.asyncio import tqdm_asyncio
API_KEY = os.environ['OPENAI_API_KEY']
BASE_URL = 'https://api.xty.app/v1'
client = AsyncOpenAI(api_key=API_KEY, base_url=BASE_URL, timeout=60.0)
file_path = '/home/qluai/zjs/TruthTorchLM-main/llama3.1/8b_results/medmcqa_llama3.1_8b_results.jsonl'

async def judge_answer(i, model_answer, ground_truths, semaphore):
    if not model_answer or not ground_truths:
        return {'line': i, 'correct': 'Incorrect'}
    prompt = f"""\nYou are a strict medical exam grader.\nGround truth answer: {ground_truths}\nModel's answer: {model_answer}\n\nQuestion: Is the model's answer semantically correct according to the ground truth? \nAnswer only with "Correct" or "Incorrect".\n"""
    async with semaphore:
        response = await client.chat.completions.create(model='gpt-4o-mini', messages=[{'role': 'system', 'content': 'You are a helpful assistant that evaluates answer correctness.'}, {'role': 'user', 'content': prompt}], temperature=0)
    judgement = response.choices[0].message.content.strip()
    return {'line': i, 'correct': judgement}

async def main():
    tasks = []
    semaphore = asyncio.Semaphore(500)
    with open(file_path, 'r', encoding='utf-8') as f:
        for (i, line) in enumerate(f, 1):
            if not line.strip():
                continue
            data = json.loads(line)
            model_answer = data.get('model_answer', '')
            ground_truths = data.get('ground_truths', [])
            if not model_answer or not ground_truths:
                tasks.append(asyncio.create_task(asyncio.sleep(0, result={'line': i, 'correct': 'Incorrect'})))
            else:
                tasks.append(judge_answer(i, model_answer, ground_truths, semaphore))
    results = []
    for coro in tqdm_asyncio.as_completed(tasks, total=len(tasks), desc='Evaluating'):
        result = await coro
        results.append(result)
    correct_count = sum((1 for r in results if r['correct'] == 'Correct'))
    incorrect_count = sum((1 for r in results if r['correct'] == 'Incorrect'))
    total = len(results)
    accuracy = correct_count / total if total else 0
    error_rate = incorrect_count / total if total else 0
    print(f'\nCorrect: {correct_count}')
    print(f'Incorrect: {incorrect_count}')
    print(f'Total judged: {total}')
    print(f'Accuracy: {accuracy:.2%}')
    print(f'Error rate: {error_rate:.2%}')
await main()
```
